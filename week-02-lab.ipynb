Run the cell. You do not need to understand the code, only the table it prints.

```python
import io
import pandas as pd

# INVENTED data for teaching. Not real measurements — do not cite it anywhere.
raw = """month,samples_taken,samples_failing,lab
January,120,7,Municipal
February,118,6,Municipal
March,121,9,Municipal
April,119,8,Municipal
May,122,31,Regional
June,120,29,Regional
July,117,33,Regional
August,0,0,Regional
September,121,30,Regional
October,119,28,Regional
"""

water = pd.read_csv(io.StringIO(raw))
water["fail_rate"] = (water["samples_failing"] / water["samples_taken"] * 100).round(1)
water
```

## Part 3 &middot; Data preparation

Preparation is not cleaning until it looks nice. It is a series of **decisions**, each of
which changes the answer, and each of which you should be able to defend.

August divided 0 failures by 0 samples. Run the cell and look carefully at the three
numbers.

```python
print("August fail_rate is:", repr(water.loc[water["month"] == "August", "fail_rate"].item()))

pandas_default = water["fail_rate"].mean()
counted_as_zero = water["fail_rate"].fillna(0).mean()
dropped_on_purpose = water.loc[water["samples_taken"] > 0, "fail_rate"].mean()

print()
print(f"Mean, however pandas felt like it:   {pandas_default:.1f}%   <- you chose nothing")
print(f"Mean, August counted as 0% failing:  {counted_as_zero:.1f}%")
print(f"Mean, August deliberately excluded:  {dropped_on_purpose:.1f}%")
```

**Answer here:**

1. `0 / 0` did not give `0`. What did `fail_rate` become for August, and what did

## Part 4 &middot; Modelling

"Modelling" does not have to mean machine learning. Here it means: pick the summary that
answers the question. Run the cell.

```python
by_lab = (water.loc[water["samples_taken"] > 0]
               .groupby("lab")["fail_rate"]
               .agg(["count", "mean", "min", "max"])
               .round(1))
by_lab
```
