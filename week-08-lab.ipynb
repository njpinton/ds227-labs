## Part 1 &middot; Correlation, in a number

**Correlation** summarises a linear relationship from &minus;1 to +1. `.corr()` does every
pair at once. Run the cell.

```python
import pandas as pd
df = pd.DataFrame({
    "study_hours": [2, 5, 1, 8, 6, 3, 7, 4],
    "sleep_hours": [8, 6, 9, 5, 6, 8, 5, 7],
    "score":       [65, 82, 60, 95, 88, 70, 90, 78],
})
print(df.corr().round(2))
```

**Answer here:**

   what does that say about their relationship?

   causation &mdash; give one reason they might move oppositely without one *causing* the

## Part 2 &middot; See it: the scatter plot

A number hides the shape; a **scatter** shows it &mdash; one point per row. Run the
cell.

```python
import pandas as pd, matplotlib.pyplot as plt
df = pd.DataFrame({
    "study_hours": [2, 5, 1, 8, 6, 3, 7, 4],
    "score":       [65, 82, 60, 95, 88, 70, 90, 78],
})
fig, ax = plt.subplots()
ax.scatter(df["study_hours"], df["score"])
ax.set_xlabel("Study hours"); ax.set_ylabel("Score"); ax.set_title("Study vs score")
plt.show()
```

**Answer here:**

1. Do the points trend upward, downward, or scatter randomly? How does that match the
   correlation number from Part 1?
   &rarr; *your answer*

2. A scatter can reveal things a correlation number cannot &mdash; a curve, a cluster, an
   outlier. Why is *seeing* the relationship worth the extra step?
   &rarr; *your answer*

## Part 3 &middot; A correlation heatmap

With many variables, a **heatmap** of the correlation matrix shows all pairs at a glance.
Run the cell.

```python
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
df = pd.DataFrame({
    "study_hours": [2, 5, 1, 8, 6, 3, 7, 4],
    "sleep_hours": [8, 6, 9, 5, 6, 8, 5, 7],
    "score":       [65, 82, 60, 95, 88, 70, 90, 78],
})
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", center=0)
plt.title("Correlation heatmap"); plt.show()
```

**Answer here:**

1. Which pair is the most strongly (positively) related, and which is the most negatively
   related? How did the colour help you spot them?
   &rarr; *your answer*

2. The diagonal is all `1.0`. Why must every variable correlate perfectly with itself?
   &rarr; *your answer*
