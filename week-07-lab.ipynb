## Part 1 &middot; Centre: mean vs median

Two ways to say 'typical'. They agree on symmetric data and diverge when it is skewed.
Run the cell.

```python
import pandas as pd
a = pd.Series([10, 12, 11, 13, 12])          # tight
b = pd.Series([10, 12, 11, 13, 120])         # one big value

for name, s in [("tight", a), ("skewed", b)]:
    print(f"{name:7} mean={s.mean():6.1f}  median={s.median():6.1f}")
```

**Answer here:**

1. For the skewed series the mean and median differ a lot. Which one moved toward the `120`,
   and which stayed near the typical value?
   &rarr; *your answer*

2. If you had to report a single 'typical' number for the skewed series, which would you
   choose, and why?

## Part 2 &middot; Spread: std and range

Centre is half the story; **spread** is the other half. Standard deviation measures how
far values sit from the mean on average. Run the cell.

```python
import pandas as pd
calm  = pd.Series([50, 51, 49, 50, 50])
wild  = pd.Series([10, 90, 30, 70, 50])

for name, s in [("calm", calm), ("wild", wild)]:
    print(f"{name:5} mean={s.mean():.0f}  std={s.std():.1f}  range={s.max()-s.min()}")
```

**Answer here:**

1. Both series have the same mean. What does the difference in `std` tell you that the mean
   alone hides?
   &rarr; *your answer*

2. Why is reporting a mean *without* a measure of spread often misleading?
   &rarr; *your answer*

## Part 3 &middot; Shape: the histogram

A histogram shows a variable's **distribution** &mdash; where it clusters and whether it
leans. Run the cell.

```python
import pandas as pd, matplotlib.pyplot as plt
income = pd.Series([15, 18, 20, 22, 19, 21, 17, 60, 16, 23, 18, 20])

fig, ax = plt.subplots()
ax.hist(income, bins=6)
ax.set_xlabel("Income (k)"); ax.set_ylabel("Count"); ax.set_title("Income distribution")
plt.show()
print("mean:", income.mean().round(1), "| median:", income.median())
```

**Answer here:**

1. The histogram has a long tail to the right (one high value). We call that
   right-skewed. How does that shape explain why the mean sits above the median here?
   &rarr; *your answer*

2. Change `bins=6` to `bins=3` and rerun. How can bin count alone change the story a
   histogram appears to tell?
   &rarr; *your answer*
