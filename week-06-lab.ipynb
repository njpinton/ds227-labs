## Part 0 &middot; Combine rows with concat

Two frames with the same columns stack with `pd.concat` &mdash; useful for joining, say,
two months of records. Run the cell.

```python
import pandas as pd
jan = pd.DataFrame({"city": ["Cebu", "Davao"], "sales": [10, 22]})
feb = pd.DataFrame({"city": ["Cebu", "Iloilo"], "sales": [14, 9]})

both = pd.concat([jan, feb], ignore_index=True)
print(both)
```

## Part 1 &middot; Integrate on a key with merge

Facts in different tables join on a shared **key**. Run the cell.

```python
import pandas as pd
sales = pd.DataFrame({"city": ["Cebu", "Davao", "Iloilo"], "sales": [24, 22, 9]})
meta  = pd.DataFrame({"city": ["Cebu", "Iloilo"], "region": ["Visayas", "Visayas"]})

merged = pd.merge(sales, meta, on="city", how="left")
print(merged)
```

**Answer here:**

1. Davao got `NaN` for `region`. Which join type produced that, and what does the `NaN`
   tell you about the match?

   behaviour you actually want?

## Part 2 &middot; Remove duplicate records

Integrating sources often duplicates rows. `drop_duplicates` keeps one of each. Run the
cell.

```python
import pandas as pd
df = pd.DataFrame({
    "city": ["Cebu", "Cebu", "Davao", "Cebu"],
    "year": [2024, 2024, 2024, 2025],
    "sales": [24, 24, 22, 30],
})
print("rows before:", len(df))
clean = df.drop_duplicates()
print("rows after: ", len(clean))
print(clean)
```

**Answer here:**

1. One row was removed. Which two rows were identical, and which row that *looks* similar
   was correctly kept?
   &rarr; *your answer*

2. `drop_duplicates(subset=["city"])` would collapse Cebu to a single row. Why could that
   be wrong here, given the `year` column?

## Part 3 &middot; Reduce to what matters

Not every column earns its place. Reducing to the relevant few makes analysis clearer and
faster. Run the cell.

```python
import pandas as pd
df = pd.DataFrame({
    "city": ["Cebu", "Davao", "Iloilo"],
    "region": ["Visayas", "Mindanao", "Visayas"],
    "sales": [24, 22, 9],
    "internal_id": ["x91", "x44", "x08"],
    "notes": ["", "", ""],
})
keep = df[["city", "region", "sales"]]
print(keep)
```
