## Part 1 &middot; Small multiples: a grid of charts

`plt.subplots(rows, cols)` lays out several axes. Each should make **one** point. Run the
cell.

```python
import pandas as pd, matplotlib.pyplot as plt
df = pd.DataFrame({
    "month":  ["Jan","Feb","Mar","Apr","May","Jun"] * 2,
    "region": ["Visayas"]*6 + ["Luzon"]*6,
    "sales":  [10,12,11,14,16,18, 20,19,22,21,25,27],
})
wide = df.pivot_table(index="month", columns="region", values="sales")

fig, (a1, a2) = plt.subplots(1, 2, figsize=(10, 3.5))
wide.plot(ax=a1, marker="o"); a1.set_title("Trend by region"); a1.set_ylabel("Sales")
df.groupby("region")["sales"].sum().plot.bar(ax=a2); a2.set_title("Total by region")
a2.set_ylabel("Sales"); plt.tight_layout(); plt.show()
```

**Answer here:**

1. The left chart shows *trend*; the right shows *totals*. Why do the two together tell a
   fuller story than either alone?
   &rarr; *your answer*

2. Each panel makes a single point. What would go wrong if you crammed both messages into
   one chart?
   &rarr; *your answer*

## Part 2 &middot; Order for the eye

A reader scans left-to-right, top-to-bottom. Put the headline first. Sorting a bar chart
guides the eye. Run the cell.

```python
import pandas as pd, matplotlib.pyplot as plt
totals = pd.Series({"Luzon": 134, "Visayas": 81, "Mindanao": 108})

fig, ax = plt.subplots()
totals.sort_values(ascending=False).plot.bar(ax=ax)
ax.set_title("Total sales by region (sorted)"); ax.set_ylabel("Sales")
plt.tight_layout(); plt.show()
```

**Answer here:**

1. Sorting put the biggest bar first. Why does that make the ranking easier to read than an
   alphabetical order?
   &rarr; *your answer*

2. On a dashboard, where should the single most important chart go, and why?
   &rarr; *your answer*

## Part 3 &middot; The caption is part of the chart

A dashboard needs words: a title that states the finding, not just the variable. Run the
cell.

```python
import matplotlib.pyplot as plt
months = ["Jan","Feb","Mar","Apr","May","Jun"]
luzon  = [20,19,22,21,25,27]

fig, ax = plt.subplots()
ax.plot(months, luzon, marker="o")
ax.set_title("Luzon sales grew 35% over the half-year")   # finding, not "Sales"
ax.set_ylabel("Sales"); plt.show()
```

**Answer here:**

1. The title states a *finding* ("grew 35%"), not just "Sales over time". Why is a
   finding-title more useful to a busy reader?
   &rarr; *your answer*

2. Write an alternative finding-title for this same chart that a reader could verify from
   the line.
   &rarr; *your answer*
