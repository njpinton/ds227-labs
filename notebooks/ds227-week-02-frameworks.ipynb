### Slide 48 — A fail-rate column and a month without data


Setup for the next cell: the small made-up data its code reads.


````python
# invented data for teaching, exactly the Week 2 lab's table
raw = """month,samples_taken,samples_failing,lab
January,120,7,Municipal
February,118,6,Municipal
March,121,9,Municipal
April,119,8,Municipal
May,122,31,Regional
June,120,29,Regional
July,117,33,Regional
August,0,0,Regional
September,121,30,Regional
October,119,28,Regional
"""
````

Expected output:

```
month  samples_taken  samples_failing        lab  fail_rate
3   April            119                8  Municipal        6.7
4     May            122               31   Regional       25.4
5    June            120               29   Regional       24.2
6    July            117               33   Regional       28.2
7  August              0                0   Regional        NaN
```


````python
import io, pandas as pd
# raw holds the ten-month table as text (see the lab)
water = pd.read_csv(io.StringIO(raw))
water["fail_rate"] = (water["samples_failing"]
                      / water["samples_taken"] * 100).round(1)
print(water.iloc[3:8])
````

### Slide 49 — Defaults as choices


Expected output:

```
16.8
15.1
```


````python
print(round(water["fail_rate"].mean(), 1))

print(round(water["fail_rate"].fillna(0).mean(), 1))
````

### Slide 50 — Writing the guard


**Exercise.**

1. Predict how many months the guard keeps, and their mean fail rate.
2. Run the cell.

````python
kept = water[water["samples_taken"] > 0]     # the guard: months with data
print(len(water), "months,", len(kept), "kept")
print(round(kept["fail_rate"].mean(), 1))
````
