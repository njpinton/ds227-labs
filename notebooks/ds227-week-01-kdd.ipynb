### Slide 44 — Loading a five-clinic table from text


Expected output:

```
clinic   barangay  visits  staff
0   C-01      Lahug   412.0      6
1   C-02     Mabolo   388.0      5
2   C-03  Guadalupe  1240.0      6
3   C-04   Talamban     NaN      4
4   C-05    Banilad   455.0      5
```


````python
import io, pandas as pd

raw = """clinic,barangay,visits,staff
C-01,Lahug,412,6
C-02,Mabolo,388,5
C-03,Guadalupe,1240,6
C-04,Talamban,,4
C-05,Banilad,455,5"""

df = pd.read_csv(io.StringIO(raw))
print(df)
````
